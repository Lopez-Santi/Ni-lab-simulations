Useful Constants

```python
GDD_fiber_permeter = 4e4 * 1e-30 #Fiber GDD as a function of length. Assume this is wavelength independent for simplicity.
GDD_CBG_852 = 4.4e8 * 1e-30
GDD_CBG_795 = 4e8 * 1e-30
modfreq_852 = 9.2e9
modfreq_795 = 6.8e9
```

Group Delay Dispersion

```python
def calc_alpha(GDD, modfreq):
    return (GDD * ((2*np.pi*modfreq)**2)/2)


print(f"Phase curvature per sideband index α for 795nm CBG: {calc_alpha(GDD_CBG_795, modfreq_795)}")
print(f"Phase curvature per sideband index α for 795nm, 5m fiber: {calc_alpha(GDD_fiber_permeter*5, modfreq_795)}")

print(f"Phase curvature per sideband index α for 852nm CBG: {calc_alpha(GDD_CBG_795, modfreq_852)}")
print(f"Phase curvature per sideband index α for 852nm, 5m fiber: {calc_alpha(GDD_fiber_permeter*5, modfreq_852)}")
```
